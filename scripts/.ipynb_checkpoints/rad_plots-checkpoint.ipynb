```python
# Cell 0 — Setup: parse RAD stats and define slowdown / T-gate helpers.
#
# Only the shared parser (aggregate_stats_by_policy) and the house color scheme
# are reused from the other notebooks. Every metric below is RAD-specific and
# computed here from scratch: SIM_CYCLES slowdown vs the ideal decoder, T-gate
# inflation, and the wrong-path FINAL_STATS groups that only RAD runs emit.
import sys, os
sys.path.insert(0, '..')
from scripts.plotting_util import aggregate_stats_by_policy
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import gmean

STATS = '../out/rad_hpca2026'
PLOTS = 'plots/rad'
os.makedirs(PLOTS, exist_ok=True)

# House color scheme (matches rdr_plots.ipynb / oclk_plots.ipynb).
C1 = '#1f4529'; C2 = '#f5f4b3'; C3 = '#abba7c'; C4 = '#898121'

# Benchmarks whose runs are not done yet: their .out files have no FINAL_STATS
# block, so the parser returns a junk dict (no SIM_CYCLES). Drop them everywhere
# until the data lands.  BQ_bose_hubbard_t is currently still running.
FILTER_OUT = ['BQ_bose_hubbard_t']

BENCHMARKS = ['BQ_bose_hubbard_q', 'BQ_bose_hubbard_t',
              'BQ_chromium_q',     'BQ_chromium_t',
              'BQ_hc3h2cn_q',      'BQ_hc3h2cn_t',
              'BQ_qaoa_3regular',  'BQ_t1B']
TICKS = ['bose_hubbard_Q', 'bose_hubbard_T',
         'chromium_Q',     'chromium_T',
         'hc3h2cn_Q',      'hc3h2cn_T',
         'qaoa_3regular',  't1B']

# Apply FILTER_OUT, keeping BENCH and TICKS_ aligned.
BENCH  = [b for b in BENCHMARKS if b not in FILTER_OUT]
TICKS_ = [t for b, t in zip(BENCHMARKS, TICKS) if b not in FILTER_OUT]

# The three RAD physical-error-rate configs, in increasing p, with plot colors.
# (C2 is a pale yellow with poor contrast, so we use C1/C3/C4 for the 3 bars.)
P_CONFIGS = [
    ('rad_p1e-06',  r'$p=10^{-6}$', C1),
    ('rad_p1e-05',  r'$p=10^{-5}$', C3),
    ('rad_p0.0001', r'$p=10^{-4}$', C4),
]

# Load stats once. ideal_baseline is the perfect/instant decoder reference.
ideal = aggregate_stats_by_policy(f'{STATS}/ideal_baseline')
rad   = {folder: aggregate_stats_by_policy(f'{STATS}/{folder}')
         for folder, _, _ in P_CONFIGS}


def slowdown_pct(rad_stats, benches):
    # Per-benchmark RAD slowdown (%) vs ideal decoder: SIM_CYCLES ratio - 1.
    return [(rad_stats[b]['SIM_CYCLES'] / ideal[b]['SIM_CYCLES'] - 1.0) * 100.0
            for b in benches]


def gmean_slowdown_pct(rad_stats, benches):
    # Geomean slowdown (%): gmean of the cycle ratio across benches, then - 1.
    ratios = [rad_stats[b]['SIM_CYCLES'] / ideal[b]['SIM_CYCLES'] for b in benches]
    return (gmean(ratios) - 1.0) * 100.0


def tgate_increase_pct(rad_stats, benches):
    # Per-benchmark extra T gates (%): T_GATES_EXECUTED / program T gates - 1.
    # The denominator (T_GATES_FROM_PROGRAM_EXECUTED) equals the ideal run's
    # T_GATES_EXECUTED, so this is exactly the wrong-path T-gate overhead.
    return [(rad_stats[b]['T_GATES_EXECUTED']
             / rad_stats[b]['T_GATES_FROM_PROGRAM_EXECUTED'] - 1.0) * 100.0
            for b in benches]


def gmean_tgate_increase_pct(rad_stats, benches):
    ratios = [rad_stats[b]['T_GATES_EXECUTED']
              / rad_stats[b]['T_GATES_FROM_PROGRAM_EXECUTED'] for b in benches]
    return (gmean(ratios) - 1.0) * 100.0


print(f'Benchmarks used ({len(BENCH)}):', BENCH)
print('Filtered out:', FILTER_OUT)
```

```python
# Cell 1 — RAD slowdown vs the ideal decoder, per workload, at three error rates.
# Grouped-bar "histogram": 3 bars (p) per workload + a trailing gmean group.
# The near-Clifford qaoa_3regular blows up at high p, so cap the y-axis and print
# the true value on top of any bar that overflows YLIM.
YLIM = 30.0

fig, ax = plt.subplots(figsize=(10.0, 2.4))

n_groups = len(BENCH) + 1            # workloads + gmean
x = np.arange(n_groups)
n_cfg = len(P_CONFIGS)
width = 0.25

for i, (folder, label, color) in enumerate(P_CONFIGS):
    heights = slowdown_pct(rad[folder], BENCH) + [gmean_slowdown_pct(rad[folder], BENCH)]
    print(f'{label:>14}: gmean slowdown = {heights[-1]:6.2f}%')
    offset = (i - n_cfg / 2 + 0.5) * width
    ax.bar(x + offset, heights, width, label=label, color=color,
           edgecolor='black', linewidth=0.5, zorder=3)
    # Label any bar that exceeds YLIM with its actual value, on top of the
    # (clipped) bar, just above the axis cap. Spread the labels within a group
    # a touch so adjacent overflow bars don't run their numbers together.
    dx = (i - (n_cfg - 1) / 2) * 0.15
    for xi, h in zip(x, heights):
        if h > YLIM:
            ax.text(xi + offset, YLIM, f'{h:.0f}%', ha='center', va='bottom',
                    fontsize=7, color='black', zorder=5, clip_on=False)

ax.set_ylim(0, YLIM)
ax.set_xticks(x)
ax.set_xticklabels(TICKS_ + ['gmean'], fontsize=10, rotation=25, ha='right')
ax.set_ylabel('Slowdown (%)', fontsize=12)
ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(len(BENCH) - 0.5, color='gray', linewidth=0.5, linestyle='--')
ax.legend(fontsize=9, ncol=1, loc='upper left', framealpha=1.0)
ax.tick_params(axis='y', labelsize=10)
ax.grid(axis='y', linestyle='-', color='gray', zorder=0)

plt.tight_layout()
plt.savefig(f'{PLOTS}/rad_slowdown.pdf', bbox_inches='tight')
plt.show()
print('Saved rad_slowdown.pdf')
```

```python
# Cell 2 — Geomean T-gate inflation (%) from wrong-path execution vs error rate.
fig, ax = plt.subplots(figsize=(5.0, 2.4))

labels = [label for _, label, _ in P_CONFIGS]
colors = [color for _, _, color in P_CONFIGS]
gms    = [gmean_tgate_increase_pct(rad[folder], BENCH) for folder, _, _ in P_CONFIGS]
for label, gval in zip(labels, gms):
    print(f'{label:>14}: gmean T-gate increase = {gval:6.3f}%')

x = np.arange(len(P_CONFIGS))
ax.bar(x, gms, 0.6, color=colors, edgecolor='black', linewidth=0.5, zorder=3)
for xi, gval in zip(x, gms):
    ax.text(xi, gval, f'{gval:.2f}%', ha='center', va='bottom', fontsize=9)

ax.set_xticks(x)
ax.set_xticklabels(labels, fontsize=11)
ax.set_xlabel('Physical Error Rate', fontsize=12)
ax.set_ylabel('T-Gate Increase (%)', fontsize=12)
ax.set_ylim(0, max(gms) * 1.25)
ax.tick_params(axis='y', labelsize=10)
ax.grid(axis='y', linestyle='-', color='gray', zorder=0)

plt.tight_layout()
plt.savefig(f'{PLOTS}/rad_tgate_increase.pdf', bbox_inches='tight')
plt.show()
print('Saved rad_tgate_increase.pdf')
```

```python
# Cell 3 — Wrong-path diagnostics by application and error rate (printout only).
# RAD-only FINAL_STATS groups: number of wrong paths, instructions executed on
# the wrong path, wrong-path latency (cycles), and qubits blocked meanwhile.
def grp(stats, group, field):
    v = stats.get(group)
    return v[field] if isinstance(v, dict) and field in v else None

def fmt(v, w=10, p=1):
    return f'{v:>{w}.{p}f}' if isinstance(v, (int, float)) else f'{"--":>{w}}'

for folder, label, _ in P_CONFIGS:
    rs = rad[folder]
    plabel = label.replace('$', '')
    print('=' * 94)
    print(f'RAD @ {plabel}   (folder: {folder})')
    print('-' * 94)
    print(f'{"workload":<16}{"#wrong":>8}{"inst_mean":>11}{"inst_max":>10}'
          f'{"lat_mean":>11}{"lat_max":>10}{"blk_mean":>10}')
    print('-' * 94)
    wp, im, lm, bm = [], [], [], []
    for b, t in zip(BENCH, TICKS_):
        s = rs[b]
        nwp   = s.get('WRONG_PATHS_DURING_EXECUTION')
        icnt  = grp(s, 'WRONG_PATH_INST_COUNT', 'MEAN')
        icmax = grp(s, 'WRONG_PATH_INST_COUNT', 'MAX')
        lat   = grp(s, 'WRONG_PATH_LATENCY',    'MEAN')
        latmx = grp(s, 'WRONG_PATH_LATENCY',    'MAX')
        blk   = grp(s, 'QUBITS_BLOCKED_DURING_WRONG_PATH', 'MEAN')
        for acc, val in ((wp, nwp), (im, icnt), (lm, lat), (bm, blk)):
            if isinstance(val, (int, float)):
                acc.append(val)
        print(f'{t:<16}{fmt(nwp,8,0)}{fmt(icnt,11)}{fmt(icmax,10,0)}'
              f'{fmt(lat,11)}{fmt(latmx,10,0)}{fmt(blk,10,2)}')
    print('-' * 94)
    print(f'{"mean":<16}{fmt(np.mean(wp),8,1)}{fmt(np.mean(im),11)}{"":>10}'
          f'{fmt(np.mean(lm),11)}{"":>10}{fmt(np.mean(bm),10,2)}')
    print()
```
