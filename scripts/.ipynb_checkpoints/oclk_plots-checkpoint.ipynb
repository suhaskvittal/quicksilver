```python
# author: Suhas Vittal
#
# Plots for CaT (out/oclk_micro2026). Run from the repo's `scripts/` directory.

import sys, os
sys.path.insert(0, 'scripts')

from plotting_util import *
import matplotlib.pyplot as plt
import numpy as np
import scipy as sp

# colors (shared with plotting.ipynb)
C1 = '#1f4529'
C2 = '#f5f4b3'
C3 = '#abba7c'
C4 = '#898121'

STATS_FOLDER = '../out/oclk_micro2026/simulation_results'
PLOT_FOLDER  = 'plots/oclk'
os.makedirs(PLOT_FOLDER, exist_ok=True)

# 5 workloads (file basenames) and how they appear on the x-axis
WORKLOADS = ['BQ_shor_rsa256_iter_4',
             'BQ_grover_3sat_schoning_1710',
             'BQ_chromium_t',
             'BQ_c2h4o_ethylene_oxide_t',
             'BQ_hc3h2cn_t']
TICKS = ['shor_rsa256', 'grover_3sat', 'chromium', 'ethylene', 'hc3h2cn']

# readout-latency reductions and the magic-state footprint budgets
RO        = [20, 30, 40, 50]
RO_COLORS = [C1, C2, C3, C4]
MSF       = [2500, 5000, 7500, 10000]   # 5000 == the no-suffix default dirs

# speedup is computed from KIPS (instruction counts are identical across configs,
# so CaT_KIPS / baseline_KIPS == baseline_cycles / cat_cycles).
STAT_KEY = 'CLIENT 0/KIPS'
```

```python
# --- plotting helpers (copied from plotting.ipynb so this notebook is self-contained) ---

def _nested_dict_access(stats: dict, key: str):
    '''Access `stats` by a '/'-separated key path, e.g. 'CLIENT 0/KIPS'.'''
    parts = key.split('/')
    d = stats
    for p in parts:
        if p not in d:
            print(f'dictionary does not have index {p}:', d.keys())
        d = d[p]
    return d


def plot_stat(ax,
                benchmark_list, benchmark_ticks, legend, configs, colors,
                stat_key, ylabel,
                ylim=None, legend_ncols=1,
                normalize_results_to_policy=None, inverted_normalization=False,
                mean_type=None, show_percentage=False):
    '''Grouped bar chart of `stat_key` for each benchmark, one bar per policy.
    See plotting.ipynb for full docs. `normalize_results_to_policy` divides every
    other policy by the named reference policy (and drops it from the plot).'''
    ax.set_ylabel(ylabel, fontsize=12)

    policy_count = len(legend)
    if normalize_results_to_policy is not None:
        policy_count -= 1
        reference_idx = legend.index(normalize_results_to_policy)

    if mean_type is not None:
        benchmark_ticks = [*benchmark_ticks, mean_type]

    benchmark_x = np.arange(len(benchmark_ticks))
    width = 1.0 / (policy_count + 1)
    offset = 0

    for (i, policy) in enumerate(legend):
        if policy == normalize_results_to_policy:
            continue
        Y = []
        for b in benchmark_list:
            y = float(_nested_dict_access(configs[i][b], stat_key))
            if normalize_results_to_policy is not None:
                reference_value = float(_nested_dict_access(configs[reference_idx][b], stat_key))
                y = reference_value / y if inverted_normalization else y / reference_value
            Y.append(y)

        if mean_type == 'amean':
            Y.append(sum(Y) / len(Y))
        elif mean_type == 'gmean':
            Y.append(sp.stats.gmean(Y))
        Y = np.array(Y)

        if show_percentage:
            Y = Y * 100 if normalize_results_to_policy is None else (Y - 1) * 100

        print(policy, Y[-1], min(Y), max(Y), Y)
        ax.bar(benchmark_x + offset, Y, width, color=colors[i], edgecolor='k', zorder=3, label=policy)
        offset += width

    if ylim is not None:
        ax.set_ylim(ylim)
        ymax = ylim[1]
        for patch in ax.patches:
            v = patch.get_height()
            if v > ymax:
                x = patch.get_x() + patch.get_width() / 2
                ax.text(x, ymax, f'{v:.3g}', ha='center', va='bottom', fontsize=10)

    workload_x_ticks = benchmark_x + (policy_count - 1) / 2 * width
    ax.set_xticks(workload_x_ticks, benchmark_ticks, rotation=30, ha='right', fontsize=10)
    ax.grid(visible=True, axis='y', zorder=0)
    if policy_count > 1:
        ax.legend(fontsize=10, ncols=legend_ncols)
```

```python
'''
    PLOT 1: CaT speedup over baseline per workload, for RO reductions 20/30/40/50.
    (default magic-state footprint, memory config)
'''

baseline = aggregate_stats_by_policy(f'{STATS_FOLDER}/baseline')
cat = [aggregate_stats_by_policy(f'{STATS_FOLDER}/cat_r{r}') for r in RO]

fig, ax = plt.subplots()
fig.set_size_inches(5.0, 2.4)
plot_stat(ax, WORKLOADS, TICKS,
          ['baseline'] + [f'{r}%' for r in RO],
          [baseline] + cat,
          [''] + RO_COLORS,
          STAT_KEY, 'Speedup',
          normalize_results_to_policy='baseline',
          mean_type='gmean')
plt.savefig(f'{PLOT_FOLDER}/oclk_speedup.pdf', bbox_inches='tight')
```

```python
'''
    PLOT 2: gmean CaT speedup vs magic-state footprint budget, one series per
    RO reduction. Each CaT config is normalized to the baseline at the SAME
    footprint budget.
'''

def _bdir(b):
    return 'baseline' if b == 5000 else f'baseline_msf{b}'

def _cdir(r, b):
    return f'cat_r{r}' if b == 5000 else f'cat_r{r}_msf{b}'

# gmean speedup[r][b] over all workloads
baselines = {b: aggregate_stats_by_policy(f'{STATS_FOLDER}/{_bdir(b)}') for b in MSF}

fig, ax = plt.subplots()
fig.set_size_inches(5.0, 2.4)
ax.set_ylabel('Speedup', fontsize=12)

msf_x = np.arange(len(MSF))
width = 1.0 / (len(RO) + 1)
offset = 0
for (i, r) in enumerate(RO):
    Y = []
    for b in MSF:
        cat_cfg = aggregate_stats_by_policy(f'{STATS_FOLDER}/{_cdir(r, b)}')
        speedups = [float(_nested_dict_access(cat_cfg[w], STAT_KEY))
                    / float(_nested_dict_access(baselines[b][w], STAT_KEY))
                    for w in WORKLOADS]
        Y.append(sp.stats.gmean(speedups))
    Y = np.array(Y)
    print(f'{r}%', Y)
    ax.bar(msf_x + offset, Y, width, color=RO_COLORS[i], edgecolor='k', zorder=3, label=f'{r}%')
    offset += width

ax.set_xticks(msf_x + (len(RO) - 1) / 2 * width, [str(b) for b in MSF], fontsize=10)
ax.grid(visible=True, axis='y', zorder=0)
ax.legend(fontsize=10)
plt.savefig(f'{PLOT_FOLDER}/oclk_msf_sensitivity.pdf', bbox_inches='tight')
```

```python
'''
    PLOT 3: CaT speedup over baseline per workload, for RO reductions 20/30/40/50,
    for the NOMEM configuration.
'''

baseline_nomem = aggregate_stats_by_policy(f'{STATS_FOLDER}/baseline_nomem')
cat_nomem = [aggregate_stats_by_policy(f'{STATS_FOLDER}/cat_r{r}_nomem') for r in RO]

fig, ax = plt.subplots()
fig.set_size_inches(5.0, 2.4)
plot_stat(ax, WORKLOADS, TICKS,
          ['baseline'] + [f'{r}%' for r in RO],
          [baseline_nomem] + cat_nomem,
          [''] + RO_COLORS,
          STAT_KEY, 'Speedup',
          normalize_results_to_policy='baseline',
          mean_type='gmean')
plt.savefig(f'{PLOT_FOLDER}/oclk_speedup_nomem.pdf', bbox_inches='tight')
```
